# 赛题任务


## 任务1：端午假期A城市交通路网拥堵识别及缓堵策略研究

### 赛题说明

伴随着城市的飞速发展，城市人口规模不断扩增，城市居民机动化出行也进入高速发展阶段。2019年A城市机动车保有量超过166万辆。虽然汽车数量的增加极大地便利了居民出行，但它所引起的负面效应同样不容忽视，城市交通拥堵现象尤为突出，尤其在节假日期间，高速公路、国省干道、城市主干道、景区周边道路等主要通道拥堵更为严重，为公众出行带来极大不便，也为交通和旅游等相关部门带来不小的管理压力。随着GPS技术在城市出租车运营管理中应用推广，交通运输部门得以获取及时准确的浮动车数据，这为道路交通运行状态评价和缓解交通拥堵治理提供数据支撑。

### 赛题任务

为更好地掌握道路交通运行状态的变化规律与趋势，参赛者需利用赛事方提供的A城市近两年端午假期出租车（含网约车）的轨迹、订单等多种数据：
- 一是完成**GPS点位地图匹配**：提供A城市交通运行指数计算方法说明、指数计算模型和指数计算结果，主要包括对比分析两年端午假期每天TOP10的拥堵路段名称、方向描述、平均运行速度（主要围绕每日早高峰时间7-9时，晚高峰17-19时）、路段拥堵延迟指数（主要围绕每日早高峰时间7-9时，晚高峰17-19时）
    
- 二是（如有）参赛者可以结合地理信息图层或其他算法软件，进一步挖掘A城市交通路网其他运行特征，提供相关的分析结论、计算结果和图层输出

- 三是针对算法输出结果，为A城市市民游客提出端午假期旅游出行建议，并结合拥堵特征提出相应地解决拥堵的策略（具体可以结合某一具体拥堵点），为解决城市拥堵治理提供辅助决策。在分析过程，参赛者必须用到但不局限于提供的数据，可自行加入自有数据进行参赛，但需说明自带数据来源并保证数据合法合规使用。

# 赛题理解

## 任务1：城市交通运行指数计算


### 道路交通运行指数

道路交通运行指数是综合反映道路网交通运行状况的指标(也称道路交通拥堵指数)，可以按照时 段设定道路交通运行指数，如日道路交通运行指数等。日道路交通运行指数可采用一日高峰时段道路交 通运行指数的平均值，主要围绕每日早高峰时间7-9时，晚高峰17-19时。

具体需要完成的细节：

- 指数计算方法说明
- 指数计算模型
- 指数计算结果

具体需要完成的内容：对比分析两年端午假期每天TOP10的拥堵路段名称、方向描述、平均运行速度（主要围绕每日早高峰时间7-9时，晚高峰17-19时）、路段拥堵延迟指数（主要围绕每日早高峰时间7-9时，晚高峰17-19时）





### 城市交通路网特征

挖掘A城市交通路网其他运行特征，提供相关的分析结论、计算结果和图层输出

任务细节：找出城市交通特点



### 端午假期旅游出行建议

为A城市市民游客提出端午假期旅游出行建议，并结合拥堵特征提出相应地解决拥堵的策略（具体可以结合某一具体拥堵点），为解决城市拥堵治理提供辅助决策。

任务细节：
- 端午出行建议：用户选择景点，进行路线推荐、时间推荐、停车推荐等出行建议；
- 拥堵解决措施：针对某一地点进行拥堵分析，并提出建议；
